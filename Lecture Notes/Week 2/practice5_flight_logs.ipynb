## One possible solution (try first!)

```python
a = pd.read_csv("flight1.csv")
b = pd.read_csv("flight2.csv")
print(a.columns.tolist())
print("max height:", a["PosZ"].max().round(2), "m;  duration:",
      round(a["Time"].iloc[-1] - a["Time"][0], 1), "s")

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].plot(a["PosX"], a["PosY"], label="Flight 1")
ax[0].plot(b["PosX"], b["PosY"], label="Flight 2")
ax[0].axis("equal"); ax[0].set_xlabel("x (m)"); ax[0].set_ylabel("y (m)"); ax[0].legend()
t = a["Time"] - a["Time"][0]   # start at 0 s
for c in ["VelX", "VelY", "VelZ"]:
    ax[1].plot(t, a[c], label=c)
ax[1].set_xlabel("time (s)"); ax[1].set_ylabel("m/s"); ax[1].legend()
plt.show()
```

**Check.** Maximum height 1.55 m, duration 59.9 s; both flights are figure-of-eight loops about 16 m wide. If your plot looks squashed, you forgot `axis("equal")`.
