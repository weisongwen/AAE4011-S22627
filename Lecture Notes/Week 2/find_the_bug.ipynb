```python
def mean_speed(xs, ys, dt):
    """Mean speed (m/s) of a drone whose position is sampled every dt seconds."""
    total = 0.0
    for k in range(len(xs)):
        dx = xs[k] - xs[k - 1]
        dy = ys[k] - ys[k - 1]
        total += (dx**2 + dy**2) ** 0.5
    return total / (len(xs) * dt)
```

## Solution (try first!)

1. `range(len(xs))` starts at `k = 0`, and `xs[-1]` is the *last* point in Python, so the loop adds a bogus segment from the end of the track back to the start.
2. A track of `N` points has `N - 1` steps, so the time is `(len(xs) - 1) * dt`, not `len(xs) * dt`.

The buggy version prints 6.667 for the test track: (10 + 5 + 5) / 3.

```python
def mean_speed_solution(xs, ys, dt):
    total = 0.0
    for k in range(1, len(xs)):
        total += ((xs[k] - xs[k - 1])**2 + (ys[k] - ys[k - 1])**2) ** 0.5
    return total / ((len(xs) - 1) * dt)

test(mean_speed_solution)
```
